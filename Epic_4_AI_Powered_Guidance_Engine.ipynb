```python
print("=" * 55)
print("EPIC 4 - STEP 1")
print("FIELD + PROBLEM + PROMPT CONSTRUCTION")
print("=" * 55)

fields = [
    "Computer Science",
    "Mathematics",
    "Physics",
    "Chemistry",
    "Biology",
    "Engineering",
    "Data Science",
    "Artificial Intelligence",
    "Business",
    "Medicine",
    "Other"
]

def build_guidance_prompt(field, problem, emotion, confidence):
    prompt = f"""
    You are an empathetic AI learning assistant.

    Student Field: {field}
    Student Problem: {problem}
    Detected Emotion: {emotion}
    Confidence: {confidence}

    Acknowledge the student's emotion.
    Give field-specific learning guidance.
    Explain the next steps clearly.
    Encourage the student.
    """
    return prompt.strip()

field = "Artificial Intelligence"
problem = "I cannot understand neural networks"
emotion = "Confused"
confidence = 0.85

prompt = build_guidance_prompt(
    field, problem, emotion, confidence
)

print("Field:", field)
print("Problem:", problem)
print("Emotion:", emotion)
print("Confidence:", confidence)

print()
print("GENERATED PROMPT:")
print(prompt)

print()
print("STEP 1 COMPLETE")
```

```python
print("=" * 55)
print("EPIC 4 - STEP 2")
print("EMPATHETIC FIELD-AWARE RESPONSE GENERATION")
print("=" * 55)

EMOTION_RESPONSES = {
    "Bored": {
        "response": "Let's make this topic more engaging with a practical example.",
        "action": "Try solving a short real-world challenge."
    },

    "Confident": {
        "response": "You are making good progress. Let's move to the next challenge.",
        "action": "Try a slightly advanced problem."
    },

    "Confused": {
        "response": "This topic can be difficult. Let's break it into smaller steps.",
        "action": "Review the basic concept and solve one simple example."
    },

    "Curious": {
        "response": "Your curiosity is valuable for learning.",
        "action": "Explore an example and connect it with what you already know."
    },

    "Frustrated": {
        "response": "Learning difficult concepts takes time. Try a simpler approach.",
        "action": "Review one concept at a time and practice gradually."
    }
}

def generate_response(field, problem, emotion, confidence, use_ai=False):

    fallback = EMOTION_RESPONSES.get(
        emotion,
        {
            "response": "Let's work through this learning problem.",
            "action": "Start with the basic concepts."
        }
    )

    return {
        "field": field,
        "problem": problem,
        "emotion": emotion,
        "confidence": confidence,
        "response": fallback["response"],
        "action": fallback["action"],
        "source": "Fallback Template" if not use_ai else "AI Simulation"
    }

result = generate_response(
    "Artificial Intelligence",
    "I cannot understand neural networks",
    "Confused",
    0.85
)

print("Generated Response:", result["response"])
print("Suggested Action:", result["action"])
print("Response Source:", result["source"])

print()
print("Fallback mechanism verified!")
print("STEP 2 COMPLETE")
```

```python
print("=" * 55)
print("EPIC 4 - STEP 3")
print("RESPONSE REGENERATION & AI TOGGLE")
print("=" * 55)

def regenerate_response(
    field,
    problem,
    emotion,
    confidence,
    use_ai
):

    result = generate_response(
        field,
        problem,
        emotion,
        confidence,
        use_ai
    )

    if use_ai:
        result["response"] = (
            f"As an AI learning assistant, I understand that "
            f"you feel {emotion.lower()} while studying {field}. "
            f"Let's approach '{problem}' step by step."
        )

        result["source"] = "AI Generated Response"

    return result


field = "Artificial Intelligence"
problem = "I cannot understand neural networks"
emotion = "Confused"
confidence = 0.85

print("TEST 1: AI OFF")

response_1 = regenerate_response(
    field,
    problem,
    emotion,
    confidence,
    False
)

print(response_1)

print()
print("TEST 2: AI ON")

response_2 = regenerate_response(
    field,
    problem,
    emotion,
    confidence,
    True
)

print(response_2)

print()
print("Emotion Score Preserved:", response_1["emotion"] == response_2["emotion"])
print("Confidence Preserved:", response_1["confidence"] == response_2["confidence"])

print()
print("Response regeneration verified!")
print("STEP 3 COMPLETE")
```

```python
import os
import pandas as pd
from datetime import datetime

print("=" * 55)
print("EPIC 4 - STEP 4")
print("SESSION HISTORY & CSV LOGGING")
print("=" * 55)

session_history = []

CSV_FILE = "emotion_response_examples.csv"


def save_interaction(result):

    interaction = {
        "field": result["field"],
        "problem": result["problem"],
        "emotion": result["emotion"],
        "confidence": result["confidence"],
        "response": result["response"],
        "action": result["action"],
        "source": result["source"],
        "timestamp": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        )
    }

    session_history.append(interaction)

    new_data = pd.DataFrame([interaction])

    if os.path.exists(CSV_FILE):

        new_data.to_csv(
            CSV_FILE,
            mode="a",
            header=False,
            index=False
        )

    else:

        new_data.to_csv(
            CSV_FILE,
            index=False
        )

    return True


test_result = regenerate_response(
    "Artificial Intelligence",
    "I cannot understand neural networks",
    "Confused",
    0.85,
    True
)

save_interaction(test_result)

saved_data = pd.read_csv(CSV_FILE)

print("Interaction saved successfully!")

print()
print("Session History Count:", len(session_history))
print("CSV Examples Count:", len(saved_data))

print()
print("LATEST SESSION:")
print(session_history[-1])

print()
print("CSV logging verified!")
print("Session history verified!")
print("Continuous learning data ready!")

print()
print("=" * 55)
print("EPIC 4 COMPLETED SUCCESSFULLY")
print("=" * 55)
```
